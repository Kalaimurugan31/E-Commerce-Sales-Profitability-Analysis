```python
# Extract the year from Order Date
import pandas as pd
import matplotlib.pyplot as plt
file_path=r"C:\Users\kalai\OneDrive\Documents\sql project\sample_-_superstore.xls"
excel_file=pd.ExcelFile(file_path)
print(excel_file.sheet_names)
df=pd.read_excel(excel_file,sheet_name="Orders")
df.head()



```

```python
# Extract the year from Order Date
df["Order Year"] = df["Order Date"].dt.year

# Count records for each year
yearly_records = df.groupby("Order Year").size()

print("Number of Records by Year:")
print(yearly_records)
```

```python
yearly_analysis = df.groupby("Order Year").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Unique_Orders=("Order ID", "nunique")
)

yearly_analysis["Profit_Margin_%"] = (
    yearly_analysis["Total_Profit"]
    / yearly_analysis["Total_Sales"]
) * 100

print(yearly_analysis.round(2))
```

```python
# Yearly sales trend
yearly_analysis["Total_Sales"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Yearly Sales Analysis")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Yearly profit trend
yearly_analysis["Total_Profit"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Yearly Profit Analysis")
plt.xlabel("Year")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Calculate yearly sales and profit growth
yearly_analysis["Sales_Growth_%"] = (
    yearly_analysis["Total_Sales"].pct_change() * 100
)

yearly_analysis["Profit_Growth_%"] = (
    yearly_analysis["Total_Profit"].pct_change() * 100
)

print(yearly_analysis.round(2))
```

```python
# Check orders dated after today

today = pd.Timestamp.today().normalize()

future_orders = df[df["Order Date"] > today]

print("Today's date:", today.date())
print("Number of future-dated orders:", len(future_orders))

if len(future_orders) > 0:
    print("Earliest future date:", future_orders["Order Date"].min())
    print("Latest future date:", future_orders["Order Date"].max())
```

```python
today = pd.Timestamp.today().normalize()

past_and_current_orders = df[df["Order Date"] <= today]
future_orders = df[df["Order Date"] > today]

print("Total records:", len(df))
print("Orders up to today:", len(past_and_current_orders))
print("Future-dated orders:", len(future_orders))

print("\nOrders by year:")
print(df["Order Date"].dt.year.value_counts().sort_index())
```

```python
# Count future-dated orders by month

today = pd.Timestamp.today().normalize()

future_orders = df[df["Order Date"] > today].copy()

future_orders["Order Month"] = future_orders["Order Date"].dt.to_period("M")

monthly_future_orders = future_orders.groupby("Order Month").size()

print("Future-dated orders by month:")
print(monthly_future_orders)

print("\nTotal future-dated records:", len(future_orders))
```

```python
# Create a dataset for analysis up to today's date

today = pd.Timestamp.today().normalize()

df_clean = df[df["Order Date"] <= today].copy()

print("Original dataset:", df.shape)
print("Analysis dataset:", df_clean.shape)

print(
    "Date range:",
    df_clean["Order Date"].min().date(),
    "to",
    df_clean["Order Date"].max().date()
)

print(
    "Future-dated orders remaining:",
    (df_clean["Order Date"] > today).sum()
)
```

```python
# Calculate key business KPIs using the cleaned dataset

total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_customers = df_clean["Customer ID"].nunique()
total_products = df_clean["Product ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Unique Products: {total_products:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
# Check duplicate records in the analysis dataset

print("Exact duplicate rows:", df_clean.duplicated().sum())

print(
    "Duplicate Order IDs:",
    df_clean["Order ID"].duplicated().sum()
)
```

```python
# Check missing values in the cleaned dataset

missing_values = df_clean.isnull().sum()

print("Missing values by column:")
print(missing_values[missing_values > 0])

print("\nTotal missing values:", df_clean.isnull().sum().sum())
```

```python
# Sales and profit by product category

category_analysis = df_clean.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).round(2)

category_analysis = category_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print(category_analysis)
```

```python
# Profitability analysis by subcategory

subcategory_analysis = df_clean.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).round(2)

subcategory_analysis = subcategory_analysis.sort_values(
    by="Total_Profit",
    ascending=True
)

print(subcategory_analysis)
```

```python
# Regional sales and profit analysis

region_analysis = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

region_analysis = region_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print(region_analysis)
```

```python
# Sales and profit by customer segment

segment_analysis = df_clean.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

segment_analysis = segment_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print(segment_analysis)
```

```python
# Discount and profitability analysis

discount_analysis = df_clean.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

discount_analysis["Profit_Margin_%"] = (
    discount_analysis["Total_Profit"]
    / discount_analysis["Total_Sales"]
) * 100

print(discount_analysis.round(2))
```

```python
# Discount vs Profit Analysis Chart

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    discount_analysis.index.astype(str),
    discount_analysis["Total_Profit"]
)

plt.axhline(y=0, color="black", linewidth=1)

plt.title("Impact of Discount Levels on Total Profit")
plt.xlabel("Discount Level")
plt.ylabel("Total Profit")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Monthly Sales and Profit Analysis

df_clean["Year_Month"] = df_clean["Order Date"].dt.to_period("M")

monthly_analysis = df_clean.groupby("Year_Month").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

print(monthly_analysis)
```

```python
# Highest and lowest sales months
highest_sales_month = monthly_analysis["Total_Sales"].idxmax()
lowest_sales_month = monthly_analysis["Total_Sales"].idxmin()

print("Highest Sales Month:", highest_sales_month)
print("Highest Sales:", monthly_analysis.loc[highest_sales_month, "Total_Sales"])

print("\nLowest Sales Month:", lowest_sales_month)
print("Lowest Sales:", monthly_analysis.loc[lowest_sales_month, "Total_Sales"])


# Highest and lowest profit months
highest_profit_month = monthly_analysis["Total_Profit"].idxmax()
lowest_profit_month = monthly_analysis["Total_Profit"].idxmin()

print("\nHighest Profit Month:", highest_profit_month)
print("Highest Profit:", monthly_analysis.loc[highest_profit_month, "Total_Profit"])

print("\nLowest Profit Month:", lowest_profit_month)
print("Lowest Profit:", monthly_analysis.loc[lowest_profit_month, "Total_Profit"])
```

```python
import matplotlib.pyplot as plt

# Convert monthly index into dates for plotting
monthly_sales = monthly_analysis.copy()
monthly_sales.index = monthly_sales.index.to_timestamp()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend (2023–2026)")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Monthly Profit Trend

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales["Total_Profit"],
    marker="o"
)

plt.axhline(y=0, color="red", linestyle="--")

plt.title("Monthly Profit Trend (2023–2026)")
plt.xlabel("Month")
plt.ylabel("Total Profit")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Yearly Sales and Profit Analysis

df_clean["Year"] = df_clean["Order Date"].dt.year

yearly_analysis = df_clean.groupby("Year").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

# Calculate yearly profit margin
yearly_analysis["Profit_Margin_%"] = (
    yearly_analysis["Total_Profit"]
    / yearly_analysis["Total_Sales"]
) * 100

# Calculate sales growth percentage
yearly_analysis["Sales_Growth_%"] = (
    yearly_analysis["Total_Sales"].pct_change() * 100
).round(2)

print(yearly_analysis.round(2))
```

```python
# Yearly Sales and Profit Comparison

yearly_analysis[["Total_Sales", "Total_Profit"]].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Yearly Sales and Profit Comparison")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend(["Total Sales", "Total Profit"])
plt.tight_layout()
plt.show()
```
